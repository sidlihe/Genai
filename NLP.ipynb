```python
import spacy

# Load the model after downloading
nlp = spacy.load("en_core_web_sm")

# Sample text processing
doc = nlp("Apple is looking at buying U.K. startup for $1 billion.")
for token in doc.ents:
    print(f'{token.text}: {token.label_}')
```

```python
doc1 = nlp("I like cats")
doc2 = nlp("I like dogs")

print(doc1.similarity(doc2))  # Returns similarity score
```

```python
text = "Apple is looking at buying U.K. startup for $1 billion."

# Process the text through the SpaCy pipeline
doc = nlp(text)

# Print named entities, their labels, and their positions in the text
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
# Import the necessary libraries
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk

# Download NLTK data (if needed)
nltk.download('punkt')

# Sample data (a few sentences)
sentences = [
    "The quick brown fox jumps over the lazy dog",
    "I love to play football",
    "Football is a great sport",
    "The lazy dog loves to sleep",
    "The brown fox is very quick and clever",
]

# Preprocess the text - tokenize the sentences into words
tokenized_sentences = [word_tokenize(sentence.lower()) for sentence in sentences]

# Train the Word2Vec model
model = Word2Vec(sentences=tokenized_sentences, vector_size=100, window=5, min_count=1, sg=1)  # sg=1 for Skip-Gram, sg=0 for CBOW

# View vector for a word
word_vector = model.wv['football']
print("Vector for 'football':\n", word_vector)

# Find similar words
similar_words = model.wv.most_similar('football', topn=5)
print("\nWords most similar to 'football':")
for word, similarity in similar_words:
    print(f"{word}: {similarity}")

# Save and load the model
model.save("word2vec_example.model")
new_model = Word2Vec.load("word2vec_example.model")

# Get similarity between two words
similarity = new_model.wv.similarity('fox', 'dog')
print(f"\nSimilarity between 'fox' and 'dog': {similarity}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample corpus (collection of documents)
documents = [
    "The quick brown fox jumps over the lazy dog",
    "Never jump over the lazy dog quickly",
    "The fox is quick and the dog is lazy",
]

# Create the TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Compute the TF-IDF matrix
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# Get the feature names (i.e., the words)
feature_names = tfidf_vectorizer.get_feature_names_out()

# Convert TF-IDF matrix to a dense format (so it's easier to read)
dense_tfidf = tfidf_matrix.todense()

# Display the TF-IDF matrix
import pandas as pd
df = pd.DataFrame(dense_tfidf, columns=feature_names)
print(df)
```
